```python
# FUNCTION: Membuat blok konvolusi fleksibel & modular
def buat_blok_konvolusi(
    in_channels, 
    out_channels, 
    kernel_size, 
    stride=1, 
    padding=0, 
    jumlah_conv=1, 
    inplace=True, 
    lrn=None, 
    pool=None  # Opsional: tuple (kernel_size, stride, padding) atau Bool
):
    layers = []
    
    # Konfigurasi LRN default jika lrn=True dikirim
    default_lrn = {
        "status": True,
        "size": 5,
        "alpha": 0.0001,
        "beta": 0.75,
        "k": 1.0
    }
    
    if lrn is True:
        lrn_config = default_lrn
    elif isinstance(lrn, dict):
        lrn_config = lrn
    else:
        lrn_config = None

    current_in = in_channels
    for i in range(jumlah_conv):
        current_out = out_channels[i] if isinstance(out_channels, (list, tuple)) else out_channels
        # Stride besar hanya diterapkan pada konvolusi pertama jika jumlah_conv > 1
        current_stride = stride if i == 0 else 1
        
        layers.append(
            nn.Conv2d(
                in_channels=current_in,
                out_channels=current_out,
                kernel_size=kernel_size,
                stride=current_stride,
                padding=padding
            )
        )
        layers.append(nn.ReLU(inplace=inplace))
        current_in = current_out
    
    # Tambahkan LRN jika dikonfigurasi dan status True
    if lrn_config and lrn_config.get("status", False):
        layers.append(
            nn.LocalResponseNorm(
                size=lrn_config.get("size", 5),
                alpha=lrn_config.get("alpha", 0.0001),
                beta=lrn_config.get("beta", 0.75),
                k=lrn_config.get("k", 1.0)
            )
        )

    # Tambahkan Max Pooling jika dikonfigurasi
    if pool:
        if isinstance(pool, tuple):
            pk = pool[0]
            ps = pool[1] if len(pool) > 1 else pk
            pp = pool[2] if len(pool) > 2 else 0
            layers.append(nn.MaxPool2d(kernel_size=pk, stride=ps, padding=pp))
        elif pool is True:
            layers.append(nn.MaxPool2d(kernel_size=3, stride=2, padding=0))

    return nn.Sequential(*layers)


class CNNBaseline(nn.Module):
    def __init__(self, num_classes=NUM_CLASSES):
        super(CNNBaseline, self).__init__()
        
        lrn_config = {
            "status": True,
            "size": 5,
            "alpha": 0.0001,
            "beta": 0.75,
            "k": 1.0
        }
        
        self.features = nn.Sequential(
            # Conv1: 96 filter (11x11, s=4, p=0) + LRN + MaxPool(3x3, s=2)
            buat_blok_konvolusi(
                in_channels=3, 
                out_channels=96, 
                kernel_size=11, 
                stride=4, 
                padding=0, 
                inplace=True, 
                lrn=lrn_config, 
                pool=(3, 2, 0)
            ),
            
            # Conv2: 256 filter (5x5, s=1, p=2) + LRN + MaxPool(3x3, s=2)
            buat_blok_konvolusi(
                in_channels=96, 
                out_channels=256, 
                kernel_size=5, 
                stride=1, 
                padding=2, 
                inplace=True, 
                lrn=lrn_config, 
                pool=(3, 2, 0)
            ),
            
            # Conv3: 384 filter (3x3, s=1, p=1) -> Tanpa LRN & Pool
            buat_blok_konvolusi(
                in_channels=256, 
                out_channels=384, 
                kernel_size=3, 
                stride=1, 
                padding=1, 
                inplace=True
            ),
            
            # Conv4: 384 filter (3x3, s=1, p=1) -> Tanpa LRN & Pool
            buat_blok_konvolusi(
                in_channels=384, 
                out_channels=384, 
                kernel_size=3, 
                stride=1, 
                padding=1, 
                inplace=True
            ),
            
            # Conv5: 256 filter (3x3, s=1, p=1) + MaxPool(3x3, s=2)
            buat_blok_konvolusi(
                in_channels=384, 
                out_channels=256, 
                kernel_size=3, 
                stride=1, 
                padding=1, 
                inplace=True, 
                pool=(3, 2, 0)
            )
        )

        # Classifier / Fully Connected Layers
        self.classifier = nn.Sequential(
            # 1. Penyesuaian Spasial & Flattening
            nn.AdaptiveAvgPool2d((6, 6)), # Memastikan output fitur selalu 7x7
            nn.Flatten(),                 # Mengubah [Batch, 512, 7, 7] menjadi [Batch, 25088]

            # 2. FC6
            nn.Linear(256 * 6 * 6, 4096),
            nn.ReLU(inplace=True),
            nn.Dropout(p=0.5),             # Regularisasi cegah overfitting

            # 3. FC7
            nn.Linear(4096, 2048),
            nn.ReLU(inplace=True),
            nn.Dropout(p=0.5),

            # 4. FC8 (Intermediat / Layer Kompresi 1000)
            nn.Linear(2048, 1000),
            nn.ReLU(inplace=True),

            # 5. FC9 (Output Logits Klasifikasi)
            nn.Linear(1000, num_classes)   # Tanpa ReLU agar nilai logits valid untuk CrossEntropyLoss
        )

    def forward(self, x):
        x = self.features(x)
        scores = self.classifier(x)   # Menghasilkan skor berukuran [N, num_classes].
        return scores  
    
```

```python
class ASPP(nn.Module):
    def __init__(
        self,
        in_channels=256,
        out_channels=256,
        branch_channels=64,
        rates=(1, 2, 3)
    ):
        super().__init__()

        # Cabang pertama: konvolusi 1 × 1.
        # Mengolah informasi kanal pada setiap posisi.
        self.cabang = nn.ModuleList([
            nn.Sequential(
                nn.Conv2d(
                    in_channels,
                    branch_channels,
                    kernel_size=1
                ),
                nn.ReLU(inplace=True)
            )
        ])

        # Cabang berikutnya: konvolusi 3 × 3
        # dengan dilation berbeda.
        for rate in rates:
            self.cabang.append(
                nn.Sequential(
                    nn.Conv2d(
                        in_channels,
                        branch_channels,
                        kernel_size=3,
                        stride=1,
                        padding=rate,
                        dilation=rate
                    ),
                    nn.ReLU(inplace=True)
                )
            )

        # Cabang global: merangkum seluruh area peta fitur.
        self.konteks_global = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Conv2d(
                in_channels,
                branch_channels,
                kernel_size=1
            ),
            nn.ReLU(inplace=True)
        )

        # Total: 1 cabang konvolusi 1 × 1,
        # 3 cabang dilated convolution, dan 1 cabang global.
        jumlah_cabang = len(rates) + 2

        # Menggabungkan kanal hasil seluruh cabang,
        # kemudian mengembalikannya menjadi 256 kanal.
        self.proyeksi = nn.Sequential(
            nn.Conv2d(
                branch_channels * jumlah_cabang,
                out_channels,
                kernel_size=1
            ),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        # Jalankan seluruh cabang konvolusi.
        hasil = [cabang(x) for cabang in self.cabang]

        # Ambil informasi global.
        global_fitur = self.konteks_global(x)

        # Samakan ukuran spasial fitur global dengan input.
        global_fitur = F.interpolate(
            global_fitur,
            size=x.shape[-2:],
            mode="bilinear",
            align_corners=False
        )

        hasil.append(global_fitur)

        # Gabungkan pada dimensi kanal.
        gabungan = torch.cat(hasil, dim=1)

        # Kembalikan jumlah kanal sesuai kebutuhan baseline.
        return self.proyeksi(gabungan)
```

```python
class CNNBaselineASPP(CNNBaseline):
    def __init__(self, num_classes=7):
        # Membentuk seluruh arsitektur CNNBaseline lama.
        super().__init__(num_classes=num_classes)

        # Pada notebook Anda, blok Conv5 berada di indeks 4.
        conv5_lama = self.features[4]

        # Susunan baru:
        # Conv5 → ReLU → ASPP → MaxPool
        self.features[4] = nn.Sequential(
            conv5_lama[0],   # Conv2d lama: 384 → 256 kanal.
            conv5_lama[1],   # ReLU lama.

            ASPP(
                in_channels=256,
                out_channels=256,
                branch_channels=64,
                rates=(1, 2, 3)
            ),

            conv5_lama[2]    # MaxPool lama.
        )

class ChannelAttention(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()

        hidden_channels = max(1, channels // reduction)

        # Ringkasan setiap kanal fitur.
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)

        # Jaringan yang sama digunakan oleh kedua ringkasan.
        self.shared_mlp = nn.Sequential(
            nn.Conv2d(
                channels,
                hidden_channels,
                kernel_size=1,
                bias=False,
            ),
            nn.ReLU(inplace=False),
            nn.Conv2d(
                hidden_channels,
                channels,
                kernel_size=1,
                bias=False,
            ),
        )

    def forward(self, x):
        avg_score = self.shared_mlp(self.avg_pool(x))
        max_score = self.shared_mlp(self.max_pool(x))

        weights = torch.sigmoid(avg_score + max_score)

        return x * weights


class SpatialAttention(nn.Module):
    def __init__(self):
        super().__init__()

        # Dua ringkasan kanal menjadi satu peta bobot spasial.
        self.conv = nn.Conv2d(
            in_channels=2,
            out_channels=1,
            kernel_size=7,
            padding=3,
            bias=False,
        )

    def forward(self, x):
        avg_map = torch.mean(x, dim=1, keepdim=True)
        max_map = torch.max(x, dim=1, keepdim=True).values

        combined = torch.cat([avg_map, max_map], dim=1)
        weights = torch.sigmoid(self.conv(combined))

        return x * weights


class CBAM(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()

        self.channel_attention = ChannelAttention(
            channels=channels,
            reduction=reduction,
        )

        self.spatial_attention = SpatialAttention()

    def forward(self, x):
        x = self.channel_attention(x)
        x = self.spatial_attention(x)

        return x


class CNNBaselineGabungan(CNNBaselineASPP):
    def __init__(
        self,
        num_classes=7,
        dilation_conv3=2,
        use_cbam=False,
    ):
        if dilation_conv3 not in (1, 2):
            raise ValueError("Gunakan dilation_conv3=1 atau 2.")

        if not isinstance(use_cbam, bool):
            raise TypeError("use_cbam harus True atau False.")

        # Membuat baseline, ASPP, dan classifier dari kelas sebelumnya.
        super().__init__(num_classes=num_classes)

        # --------------------------------------------------
        # Perubahan 1: mengatur dilation pada Conv3.
        # --------------------------------------------------
        conv3 = self.features[2][0]

        if (
            conv3.kernel_size != (3, 3)
            or conv3.stride != (1, 1)
        ):
            raise ValueError(
                "Conv3 harus memakai kernel 3x3 dan stride 1."
            )

        conv3.dilation = (dilation_conv3, dilation_conv3)
        conv3.padding = (dilation_conv3, dilation_conv3)

        # --------------------------------------------------
        # Memeriksa susunan Conv5 pada model ASPP.
        # --------------------------------------------------
        conv5 = self.features[4]

        if (
            len(conv5) != 4
            or not isinstance(conv5[2], ASPP)
            or not isinstance(conv5[3], nn.MaxPool2d)
        ):
            raise ValueError(
                "Conv5 harus tersusun "
                "Conv2d, ReLU, ASPP, MaxPool."
            )

        # --------------------------------------------------
        # Perubahan 2: memasang CBAM khusus eksperimen 2B.
        # --------------------------------------------------
        if use_cbam:
            self.features[4] = nn.Sequential(
                conv5[0],  # Conv5 lama.
                conv5[1],  # ReLU lama.
                conv5[2],  # ASPP lama.

                CBAM(
                    channels=256,
                    reduction=16,
                ),

                conv5[3],  # MaxPool lama.
            )

        # Konfigurasi untuk membentuk ulang model saat evaluasi.
        self.model_config = {
            "num_classes": int(num_classes),
            "dilation_conv3": int(dilation_conv3),
            "use_cbam": use_cbam,
        }
```

```python
num_epochs = EPOCHS
patience = PATIENCE
criterion = nn.CrossEntropyLoss()
optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE)
```

```python
def train_one_epoch(model, train_loader, criterion, optimizer, device):
    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)

        _, predicted = torch.max(outputs, 1)
        correct += (predicted == labels).sum().item()
        total += labels.size(0)

    epoch_loss = running_loss / total
    epoch_acc = correct / total

    return epoch_loss, epoch_acc
```

```python
def validate_one_epoch(model, val_loader, criterion, device):
    model.eval()

    running_loss = 0.0
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in val_loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            running_loss += loss.item() * images.size(0)

            _, predicted = torch.max(outputs, 1)
            correct += (predicted == labels).sum().item()
            total += labels.size(0)

    epoch_loss = running_loss / total
    epoch_acc = correct / total

    return epoch_loss, epoch_acc
```

```python

hasil_train = history = {
    "epoch": [],
    "train_loss": [],
    "val_loss": [],
    "train_acc": [],
    "val_acc": [],
    "timestamp": []
}

best_val_loss = float("inf")
patience_counter = 0

best_model_path = MODEL_SAVE_PATH

print("\nMemulai Pelatihan...")
for epoch in range(num_epochs):

    print(f"\n{'='*50}")
    print(f"MEMASUKI Epochs {epoch+1} | Max Epochs: {num_epochs} | Target LR: {LEARNING_RATE}")
    print(f"{'='*50}")

    train_loss, train_acc = train_one_epoch(
        model, train_loader, criterion, optimizer, device
    )

    val_loss, val_acc = validate_one_epoch(
        model, val_loader, criterion, device
    )

    history["epoch"].append(epoch + 1)
    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_acc"].append(train_acc)
    history["val_acc"].append(val_acc)
    history["timestamp"].append(datetime.now().strftime("%Y-%m-%d %H:%M:%S"))

    print(
        f"Epoch [{epoch+1}/{EPOCHS}] "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f}"
    )

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        patience_counter = 0

        torch.save({
            "epoch": epoch + 1,

            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),

            "best_val_loss": float(best_val_loss),

            "class_names": list(train_dataset.classes),
            "class_to_idx": dict(train_dataset.class_to_idx),

            "model_name": MODEL_NAME,
            "run_name": RUN_NAME,

            # Digunakan untuk membentuk ulang model saat evaluasi.
            "model_config": dict(model.model_config),
            "architecture_repr": str(model),

            "seed": int(SEED_VALUE),

            "training_config": {
                "learning_rate": float(LEARNING_RATE),
                "weight_decay": float(
                    optimizer.param_groups[0]["weight_decay"]
                ),
                "max_epochs": int(EPOCHS),
                "patience": int(PATIENCE),
                "batch_size": int(BATCH_SIZE),
                "checkpoint_metric": "val_loss",
            },

            "data_roots": {
                "train": str(train_dataset.root),
                "val": str(val_dataset.root),
                "test": str(test_dataset.root),
            },

            "data_counts": {
                "train": len(train_dataset),
                "val": len(val_dataset),
                "test": len(test_dataset),
            },
        }, best_model_path)

        print("Model terbaik disimpan.")
    else:
        patience_counter += 1
        print(f"Early stopping counter: {patience_counter}/{patience}")

    if patience_counter >= patience:
        print("Training dihentikan karena early stopping.")
        break

    print(f"{'='*50}")
```

```python
# False: evaluasi validasi untuk membandingkan konfigurasi.
# True: evaluasi test setelah konfigurasi akhir dipilih.
EVALUASI_TEST_FINAL = False

# Menggunakan checkpoint hasil training pada pelaksanaan ini.
CHECKPOINT_EVAL_PATH = MODEL_SAVE_PATH

if not CHECKPOINT_EVAL_PATH.is_file():
    raise FileNotFoundError(
        f"Checkpoint belum tersedia:\n{CHECKPOINT_EVAL_PATH}\n"
        "Jalankan training atau pilih path checkpoint sebelumnya."
    )

checkpoint = torch.load(
    CHECKPOINT_EVAL_PATH,
    map_location="cpu",
    weights_only=True,
)

# Melepaskan objek training sebelum membentuk model evaluasi.
if "optimizer" in globals():
    del optimizer

if "model" in globals():
    del model

# Membentuk ulang arsitektur sesuai konfigurasi saat training.
model = CNNBaselineGabungan(
    **checkpoint["model_config"]
)

if str(model) != checkpoint["architecture_repr"]:
    raise ValueError(
        "Definisi arsitektur berbeda dari checkpoint."
    )

model.load_state_dict(
    checkpoint["model_state_dict"],
    strict=True,
)

model = model.to(device)
model.eval()

# Memilih split evaluasi.
if EVALUASI_TEST_FINAL:
    eval_loader = test_loader
    split = "test"
    output_dir = LOG_TEST
else:
    eval_loader = val_loader
    split = "val"
    output_dir = LOG_TRAIN

# Memastikan label dataset sesuai dengan model.
if eval_loader.dataset.class_to_idx != checkpoint["class_to_idx"]:
    raise ValueError(
        "Pemetaan kelas dataset berbeda dari checkpoint."
    )

class_names = checkpoint["class_names"]
label_ids = list(range(len(class_names)))
nama_hasil = checkpoint["run_name"]

all_preds = []
all_labels = []

# Mengumpulkan prediksi.
with torch.no_grad():
    for images, labels in eval_loader:
        outputs = model(images.to(device))

        all_preds.extend(
            outputs.argmax(dim=1).cpu().tolist()
        )

        all_labels.extend(
            labels.cpu().tolist()
        )

if not all_labels:
    raise ValueError("Dataset evaluasi kosong.")

# Menghitung metrik setiap kelas.
report = classification_report(
    all_labels,
    all_preds,
    labels=label_ids,
    target_names=class_names,
    output_dict=True,
    zero_division=0,
)

print("Epoch terbaik:", checkpoint["epoch"])
print(
    "Validation loss terbaik:",
    checkpoint["best_val_loss"],
)
print("Split evaluasi:", split)

print(classification_report(
    all_labels,
    all_preds,
    labels=label_ids,
    target_names=class_names,
    digits=4,
    zero_division=0,
))

# Menyimpan classification report.
pd.DataFrame(report).T.to_csv(
    output_dir / f"{nama_hasil}_{split}_classification_report.csv"
)

# Ringkasan untuk perbandingan antareksperimen.
ringkasan = {
    "model": checkpoint["model_name"],
    "seed": checkpoint["seed"],
    "split": split,
    "jumlah_citra": len(all_labels),
    "best_epoch": checkpoint["epoch"],
    "best_val_loss": checkpoint["best_val_loss"],

    "accuracy": float(
        np.mean(
            np.array(all_labels) == np.array(all_preds)
        )
    ),

    "macro_f1": float(
        report["macro avg"]["f1-score"]
    ),

    "weighted_f1": float(
        report["weighted avg"]["f1-score"]
    ),
}

pd.DataFrame([ringkasan]).to_csv(
    output_dir / f"{nama_hasil}_{split}_metrics.csv",
    index=False,
)

print("Ringkasan:", ringkasan)

# Menghitung dan menyimpan confusion matrix.
cm = confusion_matrix(
    all_labels,
    all_preds,
    labels=label_ids,
)

pd.DataFrame(
    cm,
    index=class_names,
    columns=class_names,
).to_csv(
    output_dir / f"{nama_hasil}_{split}_confusion_matrix.csv"
)

# Menampilkan confusion matrix.
plt.figure(figsize=(9, 7))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names,
)

plt.title(
    f"{checkpoint['model_name']} - {split}"
)
plt.xlabel("Prediksi")
plt.ylabel("Label asli")
plt.tight_layout()

plt.savefig(
    output_dir / f"{nama_hasil}_{split}_confusion_matrix.png",
    dpi=300,
)

plt.show()
```
